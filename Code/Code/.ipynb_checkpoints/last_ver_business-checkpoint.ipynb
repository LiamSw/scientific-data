# BUSINESS TRAVEL AND AIR TRANSPORT ANALYSIS

## IMPORT

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.colors as mcolors
import seaborn as sns
from pathlib import Path
from textwrap import fill
from matplotlib.ticker import FuncFormatter, PercentFormatter, MaxNLocator
from matplotlib.patches import Rectangle
import requests
from itertools import product
```

## GENERAL CONFIGURATION

```python
sns.set_theme(
    style="whitegrid",
    context="notebook",
    font_scale=1.0
)

plt.rcParams.update({
    "font.family": "DejaVu Sans",
    "axes.spines.top": False,
    "axes.spines.right": False,
    "axes.titleweight": "bold",
    "axes.labelcolor": "#333333",
    "xtick.color": "#444444",
    "ytick.color": "#333333",
    "figure.dpi": 140,
    "savefig.dpi": 300,
    "savefig.bbox": "tight"
})

INDIV_FILE = "MICRODATI/VV_Microdati_2025_Individui.txt"
TRIPS_FILE = "MICRODATI/VV_Microdati_2025_Viaggi.txt"
EXCURSIONS_FILE = "MICRODATI/VV_Microdati_2025_Escursioni.txt"

OUTPUT_DIR = Path("istat_results")
OUTPUT_DIR.mkdir(exist_ok=True)
```

## LOAD AND CLEAN DATA

```python
ind = pd.read_csv(
    INDIV_FILE,
    sep="\t",
    dtype=str,
    keep_default_na=False
)

trips = pd.read_csv(
    TRIPS_FILE,
    sep="\t",
    dtype=str,
    keep_default_na=False
)

exc = pd.read_csv(
    EXCURSIONS_FILE,
    sep="\t",
    dtype=str,
    keep_default_na=False
)

# Clean column names and string values
for data in [ind, trips, exc]:
    data.columns = data.columns.str.strip()

    for col in data.select_dtypes(include="object").columns:
        data[col] = data[col].str.strip()

def convert_numeric(series):
    return pd.to_numeric(
        series.astype(str).str.replace(",", ".", regex=False),
        errors="coerce"
    )

# Convert survey weights
trips["COEV_num"] = convert_numeric(trips["COEV"])

required_cols = [
    "MOTLAV",
    "MEZZO",
    "ateco3",
    "CODPROF7",
    "COEV_num"
]

missing_cols = [
    col for col in required_cols
    if col not in trips.columns
]

if missing_cols:
    raise ValueError(
        f"Missing required variables: {missing_cols}"
    )
```

## IDENTIFY BUSINESS TRIPS AND AIR TRAVEL

```python
ATECO_LABELS = {
    "01": "Primary sector",
    "02": "Secondary sector",
    "03": "Tertiary sector"
}

PROF_LABELS = {
    "1": "Entrepreneurs and senior management",
    "2": "Knowledge and scientific professions",
    "3": "Office executive professions",
    "4": "Qualified commerce and service professions",
    "5": "Qualified manual professions",
    "6": "Unqualified manual professions",
    "7": "Armed forces"
}

MOTLAV_LABELS = {
    "01": "Congresses, conferences and seminars",
    "02": "Business meetings",
    "03": "Trade fairs and exhibitions",
    "04": "Professional or military assignments",
    "05": "Company trips and incentive travel",
    "06": "Representation, sales and technical services",
    "07": "Cultural, artistic or sporting activities",
    "08": "Teaching, consulting and training",
    "09": "Language courses",
    "10": "Vocational training",
    "11": "Transport industry personnel",
    "12": "Construction and craft activities",
    "13": "Inspection and control",
    "14": "Temporary or seasonal employment",
    "15": "Other business purposes"
}

for col in ["MOTLAV", "MEZZO", "ateco3", "CODPROF7"]:
    trips[col] = trips[col].astype(str).str.strip()

business_trips = trips[
    trips["MOTLAV"].ne("")
].copy()

business_trips["is_air"] = (
    business_trips["MEZZO"] == "01"
)

business_trips = business_trips[
    business_trips["COEV_num"].notna() &
    (business_trips["COEV_num"] > 0)
].copy()

if business_trips.empty:
    raise ValueError(
        "No valid business-trip observations"
    )

# General indicators
n_total = len(business_trips)
n_air = int(business_trips["is_air"].sum())
weighted_total = business_trips["COEV_num"].sum()
weighted_air = business_trips.loc[
    business_trips["is_air"], "COEV_num"
].sum()

weighted_share = (
    weighted_air / weighted_total * 100
    if weighted_total > 0 else np.nan
)
```

## WEIGHTED ANALYSIS FUNCTION

```python
def weighted_ranking(data, variable, labels):
    """
    Calculate observed sample sizes, weighted trip volumes
    and weighted air-travel shares for each category.
    """

    temp = data[
        data[variable].notna() &
        data[variable].astype(str).str.strip().ne("")
    ].copy()

    temp[variable] = temp[variable].astype(str).str.strip()

    rows = []

    for code, group in temp.groupby(variable):

        n_sample = len(group)

        n_air_sample = int(group["is_air"].sum())

        # Estimated total number of business trips
        total_est = group["COEV_num"].sum()

        # Estimated number of business trips by air
        air_est = group.loc[
            group["is_air"], "COEV_num"
        ].sum()

        # Weighted share of air travel
        share = (
            air_est / total_est * 100
            if total_est > 0 else np.nan
        )

        rows.append({
            "code": code,
            "Group": labels.get(code, f"Code {code}"),
            "n_sample": n_sample,
            "n_air_sample": n_air_sample,
            "trips_est": total_est,
            "air_trips_est": air_est,
            "air_share": share
        })

    result = pd.DataFrame(rows)

    if result.empty:
        return result

    # Indicative sample-size warning
    result["Sample_size_warning"] = np.where(
        result["n_sample"] < 20,
        "Small sample (<20)",
        "20 or more observations"
    )

    return result.sort_values(
        "air_trips_est",
        ascending=False
    ).reset_index(drop=True)
```

## CALCULATE RESULTS BY SECTOR, PROFESSION AND MOTIVE

```python
ranking_ateco3 = weighted_ranking(
    business_trips,
    "ateco3",
    ATECO_LABELS
)

ranking_codprof7 = weighted_ranking(
    business_trips,
    "CODPROF7",
    PROF_LABELS
)

ranking_motives = weighted_ranking(
    business_trips,
    "MOTLAV",
    MOTLAV_LABELS
)
```

## VISUALISATION FUNCTION

```python
plt.rcParams.update({
    "font.family": "serif",
    "axes.spines.top": False,
    "axes.spines.right": False,
    "axes.grid": False,
    "figure.dpi": 150
})

HEATMAP_CMAP = mcolors.LinearSegmentedColormap.from_list(
    "air_blues", ["#f4f6f7", "#aed6f1", "#2e86c1", "#154360"]
)
HEATMAP_NORM = mcolors.Normalize(vmin=0, vmax=85)


# Bar chart
def plot_ranking_classic(data, title, filename):
    if data.empty:
        print(f"No data available for: {title}")
        return

    df_plot = data.sort_values("air_trips_est", ascending=True).copy()
    df_plot["label"] = df_plot["Group"].apply(lambda x: fill(str(x), width=36))

    fig_height = max(4.2, len(df_plot) * 0.65)
    fig, (ax1, ax2) = plt.subplots(
        1, 2,
        figsize=(14.5, fig_height),
        sharey=True,
        gridspec_kw={"width_ratios": [1.15, 1.0], "wspace": 0.12}
    )

    y_pos = list(range(len(df_plot)))

    # Estimated volume of business trips by air
    ax1.barh(
        y_pos,
        df_plot["air_trips_est"],
        height=0.56,
        color="#1f618d",
        edgecolor="none"
    )
    ax1.set_yticks(y_pos)
    ax1.set_yticklabels(df_plot["label"], fontsize=10)
    ax1.set_title("Estimated business trips by air", loc="left", fontsize=11, fontweight="bold", pad=26)
    ax1.xaxis.set_major_formatter(FuncFormatter(lambda x, _: f"{x:,.0f}"))
    ax1.grid(False)
    ax1.tick_params(axis="y", length=0)
    ax1.tick_params(axis="x", labelsize=9)

    max_vol = df_plot["air_trips_est"].max()
    ax1.set_xlim(0, max_vol * 1.24 if max_vol > 0 else 1)

    for y, val in zip(y_pos, df_plot["air_trips_est"]):
        ax1.text(
            val + (max_vol * 0.018),
            y,
            f"{val:,.0f}",
            va="center",
            ha="left",
            fontsize=8.5,
            color="#1b2631"
        )

    # Share of business trips made by air
    air_shares = df_plot["air_share"].fillna(0).values
    other_shares = 100.0 - air_shares

    ax2.barh(
        y_pos,
        air_shares,
        height=0.56,
        color="#1f618d",
        edgecolor="white",
        linewidth=1,
        label="Air"
    )
    ax2.barh(
        y_pos,
        other_shares,
        left=air_shares,
        height=0.56,
        color="#eaecee",
        edgecolor="white",
        linewidth=1,
        label="Other modes"
    )

    ax2.set_title("Share of business trips made by air", loc="left", fontsize=11, fontweight="bold", pad=26)
    ax2.set_xlim(0, 100)
    ax2.xaxis.set_major_formatter(PercentFormatter(xmax=100, decimals=0))
    ax2.grid(False)
    ax2.tick_params(axis="y", length=0)
    ax2.tick_params(axis="x", labelsize=9)
    ax2.legend(ncol=2, loc="lower left", bbox_to_anchor=(0.0, 1.01), frameon=False, fontsize=8.5)

    for y, val in zip(y_pos, air_shares):
        if val >= 0:
            ax2.text(
                val - 2.2,
                y,
                f"{val:.1f}%",
                va="center",
                ha="right",
                fontsize=8.5,
                fontweight="bold",
                color="white"
            )
        else:
            ax2.text(
                val + 2.0,
                y,
                f"{val:.1f}%",
                va="center",
                ha="left",
                fontsize=8.5,
                fontweight="bold",
                color="#1b2631"
            )

    fig.suptitle(title, x=0.01, y=1.06, ha="left", fontsize=13.5, fontweight="bold")
    fig.savefig(OUTPUT_DIR / filename, dpi=300, bbox_inches="tight")
    plt.show()


# Bar chart with heatmap-style
def plot_ranking(data, title, filename):
    if data.empty:
        print(f"No data available for: {title}")
        return

    df_plot = data.sort_values("air_trips_est", ascending=True).copy()
    df_plot["label"] = df_plot["Group"].apply(lambda x: fill(str(x), width=36))

    n_rows = len(df_plot)
    fig_height = max(4.8, n_rows * 0.50)

    fig, (ax1, ax2) = plt.subplots(
        1, 2,
        figsize=(13.5, fig_height),
        gridspec_kw={"width_ratios": [3.3, 1.0], "wspace": 0.05}
    )

    y_pos = np.arange(n_rows)
    bar_h = 0.58
    tile_h = 0.66

    # Estimated volume of business trips by air (bar chart)
    ax1.barh(
        y_pos,
        df_plot["air_trips_est"],
        height=bar_h,
        color="#1f618d",
        edgecolor="none",
        zorder=3
    )

    max_vol = df_plot["air_trips_est"].max()
    ax1.set_xlim(0, max_vol * 1.18 if max_vol > 0 else 1)
    ax1.set_ylim(-0.6, n_rows - 0.35)
    ax1.set_yticks(y_pos)
    ax1.set_yticklabels(df_plot["label"], fontsize=9.5)
    ax1.tick_params(axis="y", length=0)
    ax1.tick_params(axis="x", labelsize=9.3)
    ax1.xaxis.set_major_locator(MaxNLocator(nbins=5, integer=True))
    ax1.xaxis.set_major_formatter(FuncFormatter(lambda x, _: f"{x:,.0f}"))
    ax1.grid(False)
    ax1.set_title(
        "Estimated business trips by air",
        loc="left",
        fontsize=11,
        fontweight="bold",
        pad=12
    )

    for y, val in zip(y_pos, df_plot["air_trips_est"]):
        ax1.text(
            val + (max_vol * 0.016 if max_vol > 0 else 0.02),
            y,
            f"{val:,.0f}",
            va="center",
            ha="left",
            fontsize=9.0,
            color="#1b2631"
        )

    # Share of business trips made by air (heatmap-style)
    air_shares = df_plot["air_share"].fillna(0).values

    for y, val in zip(y_pos, air_shares):
        rect = Rectangle(
            (0.08, y - tile_h / 2),
            0.84,
            tile_h,
            facecolor=HEATMAP_CMAP(HEATMAP_NORM(val)),
            edgecolor="white",
            linewidth=1.3
        )
        ax2.add_patch(rect)

        ax2.text(
            0.50,
            y,
            f"{val:.1f}%",
            va="center",
            ha="center",
            fontsize=9.2,
            fontweight="bold",
            color="white" if val >= 40.0 else "#1b2631"
        )

    ax2.set_xlim(0, 1)
    ax2.set_ylim(-0.6, n_rows - 0.35)
    ax2.set_yticks([])
    ax2.set_xticks([])
    for spine in ["top", "right", "left", "bottom"]:
        ax2.spines[spine].set_visible(False)

    ax2.set_title(
        "Share of business\ntrips made by air",
        loc="center",
        fontsize=10.5,
        fontweight="bold",
        pad=8
    )

    fig.suptitle(
        title,
        x=0.01,
        y=1.01,
        ha="left",
        fontsize=13.5,
        fontweight="bold"
    )
    fig.savefig(OUTPUT_DIR / filename, dpi=300, bbox_inches="tight")
    plt.show()
```

##  EUROSTAT — AIR TRAVEL SHARE BY PURPOSE OF TRAVEL

```python
EUROSTAT_FILE = "tour_dem_tttr$defaultview_linear_2_0.csv"
euro = pd.read_csv(EUROSTAT_FILE)[
    ["geo", "c_dest", "purpose", "duration", "tra_mode", "TIME_PERIOD", "OBS_VALUE"]
].copy()

euro["OBS_VALUE"] = pd.to_numeric(euro["OBS_VALUE"], errors="coerce")
euro["TIME_PERIOD"] = pd.to_numeric(euro["TIME_PERIOD"], errors="coerce")
euro = euro.dropna(subset=["OBS_VALUE", "TIME_PERIOD"]).copy()
euro["TIME_PERIOD"] = euro["TIME_PERIOD"].astype(int)

latest_year = euro["TIME_PERIOD"].max()

sub = euro[
    (euro["geo"] == "EU27_2020") &
    (euro["c_dest"] == "WORLD") &
    (euro["duration"] == "N_GE1") & # N_GE1 = 1 night or more
    (euro["TIME_PERIOD"] == latest_year) &
    (euro["purpose"].isin(["PER", "PROF"]))
].copy()

piv = sub.pivot_table(index="purpose", columns="tra_mode", values="OBS_VALUE", aggfunc="sum")

# Calculate total trips for each purpose
modal = pd.DataFrame(index=piv.index)
modal["Air"] = piv["AIR"] / piv["TOTAL"] * 100
modal["Rail"] = piv["RAIL"] / piv["TOTAL"] * 100
modal["Car & road"] = piv["RDMV_PAS_X_BUS"] / piv["TOTAL"] * 100
modal["Bus & other"] = 100 - (modal["Air"] + modal["Rail"] + modal["Car & road"])

PURPOSE_LABELS = {
    "PER": "Personal trips",
    "PROF": "Business trips"
}
modal.index = modal.index.map(PURPOSE_LABELS)
modal = modal.loc[[PURPOSE_LABELS["PER"], PURPOSE_LABELS["PROF"]]]

# Calculate total volume of air travel by purpose
air_vol = piv["AIR"]
air_vol_share = (air_vol / air_vol.sum() * 100).rename(index=PURPOSE_LABELS)

# Visualization
plt.rcParams.update({
    "font.family": "serif",
    "axes.spines.top": False,
    "axes.spines.right": False,
    "axes.grid": False,
    "figure.dpi": 150
})

fig, (ax1, ax2) = plt.subplots(
    2, 1,
    figsize=(10.5, 5.6),
    gridspec_kw={"height_ratios": [2.2, 1.0], "hspace": 0.72}
)

# Modal part of the travel purpose
colors_modes = ["#1f618d", "#85929e", "#bfc9ca", "#e5e8e8"]
modes = ["Air", "Rail", "Car & road", "Bus & other"]

left = np.zeros(len(modal))
y_pos = np.arange(len(modal))

for mode, color in zip(modes, colors_modes):
    vals = modal[mode].values
    ax1.barh(y_pos, vals, left=left, height=0.55, color=color, edgecolor="white", linewidth=1.2, label=mode)
    
    for i, (val, l_pos) in enumerate(zip(vals, left)):
        if val >= 3.0:
            txt_color = "white" if mode == "Air" else "#2c3e50"
            fw = "bold" if mode == "Air" else "normal"
            label_txt = f"Air: {val:.1f}%" if mode == "Air" else f"{val:.1f}%"
            ax1.text(l_pos + val / 2, y_pos[i], label_txt, ha="center", va="center", color=txt_color, fontsize=9.5, fontweight=fw)
    left += vals

ax1.set_yticks(y_pos)
ax1.set_yticklabels(modal.index, fontsize=10)
ax1.set_xlim(0, 100)
ax1.xaxis.set_major_formatter(PercentFormatter(xmax=100, decimals=0))
ax1.set_xlabel("Modal share within each travel purpose", fontsize=9.5)
ax1.set_title(f"Modal Split by Travel Purpose in Europe ({latest_year})", loc="left", fontsize=11.5, fontweight="bold", pad=28)
ax1.legend(ncol=4, loc="upper left", bbox_to_anchor=(0, 1.15), frameon=False, fontsize=9)
ax1.grid(False)

# Total volume of air travel by purpose
air_biz = air_vol_share[PURPOSE_LABELS["PROF"]]
air_per = air_vol_share[PURPOSE_LABELS["PER"]]

ax2.barh([0], [air_biz], left=0, height=0.45, color="#1f618d", edgecolor="white", linewidth=1.2)
ax2.barh([0], [air_per], left=air_biz, height=0.45, color="#7fb3d5", edgecolor="white", linewidth=1.2)

ax2.text(air_biz / 2, 0, f"Business\n{air_biz:.1f}%", ha="center", va="center", color="white", fontsize=9.5, fontweight="bold")
ax2.text(air_biz + air_per / 2, 0, f"Personal trips: {air_per:.1f}%", ha="center", va="center", color="#1b2631", fontsize=9.5, fontweight="bold")

ax2.set_yticks([])
ax2.set_xlim(0, 100)
ax2.xaxis.set_major_formatter(PercentFormatter(xmax=100, decimals=0))
ax2.set_xlabel("Share of total air travel volume", fontsize=9.5)
ax2.set_title(f"Composition of Total Air Travel Volume by Purpose in Europe({latest_year})", loc="left", fontsize=11.5, fontweight="bold", pad=12)
ax2.grid(False)

fig.savefig(OUTPUT_DIR / "eurostat_air_travel_by_purpose_paired.png", dpi=300, bbox_inches="tight")
plt.show()
```

## FREQUENCY INEQUALITY

```python
DATASET = "ilc_enevt_05a"
API_URL = (
    "https://ec.europa.eu/eurostat/api/dissemination/"
    f"statistics/1.0/data/{DATASET}"
)

params = {
    "lang": "en",
    "geo": "IT",
    "freq": "A",
    "unit": "PC",
    "time": "2025",
    "yn_arope": "TOTAL",
    "deg_urb": "TOTAL",
    "hhcomp": "TOTAL"
}

response = requests.get(API_URL, params=params, timeout=60)
response.raise_for_status()
data = response.json()

# Parse JSON response
dimension_ids = data["id"]
dimension_sizes = data["size"]
dimensions = data["dimension"]

# Convert the JSON values into a long-format DataFrame.
records = []
values = data.get("value", [])
statuses = data.get("status", [])

for coordinates in product(
    *[range(size) for size in dimension_sizes]
):
    flat_index = 0

    for i, coordinate in enumerate(coordinates):
        multiplier = int(np.prod(dimension_sizes[i + 1:])) if i + 1 < len(dimension_sizes) else 1
        flat_index += coordinate * multiplier

    if isinstance(values, dict):
        value = values.get(str(flat_index))
    else:
        value = values[flat_index] if flat_index < len(values) else None

    if value is None:
        continue

    record = {"OBS_VALUE": float(value)}

    for i, dimension_id in enumerate(dimension_ids):
        category = dimensions[dimension_id]["category"]
        category_index = category["index"]

        reverse_index = {
            position: code
            for code, position in category_index.items()
        }

        code = reverse_index[coordinates[i]]
        labels = category.get("label", {})
        record[dimension_id] = code
        record[f"{dimension_id}_LABEL"] = labels.get(code, code)

    if isinstance(statuses, list) and flat_index < len(statuses):
        record["OBS_STATUS"] = statuses[flat_index]
    elif isinstance(statuses, dict):
        record["OBS_STATUS"] = statuses.get(str(flat_index), "")
    else:
        record["OBS_STATUS"] = ""

    records.append(record)

euro = pd.DataFrame(records)

if euro.empty:
    raise ValueError("No Eurostat observations were returned.")


# Select the three frequency categories

# Eurostat dimension name returned by the API: "frequenc"
frequency_col = "frequenc_LABEL"

if frequency_col not in euro.columns:
    raise ValueError(
        f"Frequency dimension not found. Available columns: {euro.columns.tolist()}"
    )

frequency_labels = {
    "Not in the last 12 months": "No flights",
    "Once (one time)": "Once",
    "2 times or more": "2 times or more"
}

euro["frequency"] = euro[frequency_col].map(frequency_labels)

# Check for unexpected or unmapped categories
unmapped = euro.loc[
    euro[frequency_col].notna() & euro["frequency"].isna(),
    frequency_col
].unique()

if len(unmapped) > 0:
    raise ValueError(
        f"Unexpected frequency labels: {unmapped}. "
    )

euro = euro[euro["frequency"].notna()].copy()

# Check that exactly three categories are available
if euro["frequency"].duplicated().any() or len(euro) != 3:
    raise ValueError(
        "Expected exactly three frequency observations. "
        f"{euro[['frequency', 'OBS_VALUE']].to_string(index=False)}"
    )

frequency_order = [
    "No flights",
    "Once",
    "2 times or more"
]

euro["frequency"] = pd.Categorical(
    euro["frequency"],
    categories=frequency_order,
    ordered=True
)

euro = euro.sort_values("frequency").reset_index(drop=True)


# Calculate participation and frequency indicators

no_flight_pct = float(
    euro.loc[
        euro["frequency"] == "No flights",
        "OBS_VALUE"
    ].iloc[0]
)

once_pct = float(
    euro.loc[
        euro["frequency"] == "Once",
        "OBS_VALUE"
    ].iloc[0]
)

two_plus_pct = float(
    euro.loc[
        euro["frequency"] == "2 times or more",
        "OBS_VALUE"
    ].iloc[0]
)

# Share of the population that took at least one flight
active_flyers_pct = once_pct + two_plus_pct

# Frequency distribution among active air travellers
if active_flyers_pct == 0:
    raise ValueError("No active air travellers found in the selected data.")

once_active_pct = (once_pct / active_flyers_pct) * 100
two_plus_active_pct = (two_plus_pct / active_flyers_pct) * 100

# Data for export

results = pd.DataFrame({
    "Frequency": [
        "No flights",
        "Once",
        "2 times or more"
    ],
    "Share of total population (%)": [
        no_flight_pct,
        once_pct,
        two_plus_pct
    ],
    "Share among active air travellers (%)": [
        np.nan,
        once_active_pct,
        two_plus_active_pct
    ]
})

# Visualisation

plt.rcParams.update({
    "font.family": "serif",
    "axes.grid": False,
    "figure.dpi": 150
})

fig, (ax1, ax2) = plt.subplots(
    2, 1,
    figsize=(10.0, 4.5),
    gridspec_kw={
        "height_ratios": [1.0, 1.0],
        "hspace": 0.85
    }
)

# Participation in the total population

ax1.barh(
    [0],
    [active_flyers_pct],
    height=0.46,
    color="#1f618d",
    edgecolor="white",
    linewidth=1.2
)

ax1.barh(
    [0],
    [no_flight_pct],
    left=[active_flyers_pct],
    height=0.46,
    color="#eaecee",
    edgecolor="white",
    linewidth=1.2
)

ax1.text(
    active_flyers_pct / 2,
    0,
    f"Active flyers\n{active_flyers_pct:.1f}%",
    ha="center",
    va="center",
    fontsize=9.5,
    color="white",
    fontweight="bold"
)

ax1.text(
    active_flyers_pct + no_flight_pct / 2,
    0,
    f"No flights: {no_flight_pct:.1f}%",
    ha="center",
    va="center",
    fontsize=9.5,
    color="#2c3e50",
    fontweight="bold"
)

ax1.set_xlim(0, 100)
ax1.set_ylim(-0.5, 0.5)
ax1.set_yticks([0])
ax1.set_yticklabels(["Total population\n(aged 16+)"], fontsize=9.5)
ax1.xaxis.set_major_formatter(PercentFormatter(xmax=100, decimals=0))
ax1.set_xlabel("Share of total population", fontsize=9.5)
ax1.set_title(
    "Annual Air Travel Participation in Italy (2025)",
    loc="left",
    fontsize=11.5,
    fontweight="bold",
    pad=12
)

# Frequency among active travellers TRAVELLERS

ax2.barh(
    [0],
    [once_active_pct],
    height=0.46,
    color="#7fb3d5",
    edgecolor="white",
    linewidth=1.2
)

ax2.barh(
    [0],
    [two_plus_active_pct],
    left=[once_active_pct],
    height=0.46,
    color="#1f618d",
    edgecolor="white",
    linewidth=1.2
)

ax2.text(
    once_active_pct / 2,
    0,
    f"Once (1 flight)\n{once_active_pct:.1f}%",
    ha="center",
    va="center",
    fontsize=9.5,
    color="#1b2631",
    fontweight="bold"
)

ax2.text(
    once_active_pct + two_plus_active_pct / 2,
    0,
    f"2 times or more\n{two_plus_active_pct:.1f}%",
    ha="center",
    va="center",
    fontsize=9.5,
    color="white",
    fontweight="bold"
)

ax2.set_xlim(0, 100)
ax2.set_ylim(-0.5, 0.5)
ax2.set_yticks([])
ax2.xaxis.set_major_formatter(PercentFormatter(xmax=100, decimals=0))
ax2.set_xlabel("Share among active air travellers", fontsize=9.5)
ax2.set_title(
    "Flight Frequency Among Active Flyers (Previous 12 Months)",
    loc="left",
    fontsize=11.5,
    fontweight="bold",
    pad=12
)

for ax in (ax1, ax2):
    ax.spines["top"].set_visible(False)
    ax.spines["right"].set_visible(False)
    ax.tick_params(axis="y", length=0)
    ax.grid(False)

# Save outputs

OUTPUT_DIR = Path(OUTPUT_DIR)
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

fig.savefig(
    OUTPUT_DIR / "eurostat_air_travel_participation_frequency_2025.png",
    dpi=300,
    bbox_inches="tight"
)

results.to_csv(
    OUTPUT_DIR / "eurostat_air_travel_participation_frequency_2025.csv",
    index=False,
    encoding="utf-8-sig"
)

plt.show()
```
